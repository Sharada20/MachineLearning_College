```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# ---- Load dataset ----
df = pd.read_csv("/content/iris.csv")

# Map text labels to target integers
class_names = sorted(df['species'].unique())
label_mapping = {name: idx for idx, name in enumerate(class_names)}
y = df['species'].map(label_mapping)

# Features are all columns except 'species'
X = df.drop(columns=['species'])
feature_names = X.columns.tolist()

# ---- Train-test split ----
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# =========================================================
# CART using Gini impurity (default in sklearn's DecisionTreeClassifier)
# =========================================================
cart_gini = DecisionTreeClassifier(criterion='gini', max_depth=4, random_state=42)
cart_gini.fit(X_train, y_train)
y_pred_gini = cart_gini.predict(X_test)

print("=" * 60)
print("CART (Gini impurity)")
print("=" * 60)
print(f"Accuracy: {accuracy_score(y_test, y_pred_gini):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_gini, target_names=class_names))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_gini))

print("\nFeature importances:")
for name, importance in zip(feature_names, cart_gini.feature_importances_):
    print(f"  {name}: {importance:.4f}")

# =========================================================
# CART using entropy criterion (for comparison)
# =========================================================
cart_entropy = DecisionTreeClassifier(criterion='entropy', max_depth=4, random_state=42)
cart_entropy.fit(X_train, y_train)
y_pred_entropy = cart_entropy.predict(X_test)

print("\n" + "=" * 60)
print("CART (Entropy / Information Gain) - for comparison")
print("=" * 60)
print(f"Accuracy: {accuracy_score(y_test, y_pred_entropy):.4f}")

# ---- Predict a new sample ----
new_sample = pd.DataFrame([[5.1, 3.5, 1.4, 0.2]], columns=feature_names)
predicted_class = cart_gini.predict(new_sample)[0]
print(f"\nPrediction for new sample {new_sample.values.tolist()}: {class_names[predicted_class]}")

# ---- Visualize the decision tree ----
plt.figure(figsize=(16, 10))
plot_tree(cart_gini, feature_names=feature_names, class_names=class_names,
          filled=True, rounded=True, fontsize=9)
plt.title("CART Decision Tree (Gini) - Iris Dataset")
plt.tight_layout()
plt.savefig("exp8_decision_tree.png", dpi=150)
print("\nDecision tree plot saved as exp8_decision_tree.png")
```
